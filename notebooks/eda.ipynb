# Component 3 — EDA and ablation figures

Load window-level visual features, inspect class balance and validity ratios, and (once trained) read `reports/baseline_cv.json` / `reports/ablation.json` for thesis tables.

Run `python -m component3 synthetic` first if you do not yet have participant data.

```python
from pathlib import Path
import json
import pandas as pd

root = Path("..") if Path("../config/default.yaml").exists() else Path(".")
feat_path = root / "data/processed/features/windows.parquet"
if feat_path.exists():
    df = pd.read_parquet(feat_path)
    display(df.head())
    print("windows", len(df), "participants", df["participant_id"].nunique())
    print(df[["face_present_ratio", "gaze_valid_ratio", "off_screen_gaze_ratio", "phone_present_ratio"]].describe())
else:
    print("No windows.parquet yet — run: python -m component3 synthetic")
```

```python
for name in ("baseline_cv.json", "evaluation.json", "ablation.json", "benchmark.json"):
    p = root / "reports" / name
    print("==", name, "==")
    if p.exists():
        print(json.dumps(json.loads(p.read_text()), indent=2)[:1500])
    else:
        print("(missing)")
```
